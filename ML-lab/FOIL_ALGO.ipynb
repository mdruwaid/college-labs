```python
import pandas as pd
!pip install Orange3
from Orange.data import Table, Domain, DiscreteVariable
from Orange.classification.rules import CN2Learner
```

```python

df = pd.read_csv(r"zoo.csv")
```

```python

target_column = "class_type"
non_feature_columns = ["animal_name"] if "animal_name" in df.columns else []
```

```python

X= df.drop(non_feature_columns + [target_column], axis=1)
y= df[target_column]
```

```python

X = X.astype(str)
y = y.astype(str)
```

```python

X = X.astype(str)
y = y.astype(str)
```

```python

feature_vars = []
for col in X.columns:
  values = sorted(X[col].unique().tolist())
  feature_vars.append(DiscreteVariable(col, values=values))
```

```python
target_values = sorted(y.unique().tolist())
target_var = DiscreteVariable(target_column, values=target_values)
```

```python
domain = Domain (feature_vars, target_var)
```

```python

data_orange = Table.from_list(domain, [list(row) + [label] for row, label in zip(X.values.tolist(), y.tolist())])


learner = CN2Learner()
classifier = learner(data_orange)


print("📜 CN2 Classification Rules:")
for rule in classifier.rule_list:
    print(rule)
```

```python
from pyswip import Prolog
prolog = Prolog()
prolog.assertz("parent(john, mary)")
print(list(prolog.query("parent(john, mary)")))
prolog.assertz("has_hair(cat)")
prolog.assertz("gives_milk(cat)")
prolog.assertz("has_backbone(cat)")
prolog.assertz("class(cat, mammal)")
prolog.assertz("has_feathers(penguin)")
prolog.assertz("lays_eggs(penguin)")
prolog.assertz("class(penguin, bird)")
print("\n FOIL-style Rule Queries:")
for result in prolog.query("class(X, mammal)"):
    print("Mammal:", result["X"])

for result in prolog.query("class(X, bird)"):
    print("Bird:", result["X"])
```
