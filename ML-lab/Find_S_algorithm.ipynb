```python
def finds(examples,target):
  hypothesis = examples[0].copy()
  for i,val in enumerate(examples):
    if target[i] == "yes":
      for x in range(len(hypothesis)):
        if hypothesis[x] != val[x]:
          hypothesis[x] = "?"
  return hypothesis
examples = [["sunny","warm","normal","strong","warm","same"],
            ["sunny","warm","high","strong","warm","same"],
            ["sunny","warm","high","strong","cool","change"],
            ["rainy","cold","high","strong","warm","change"]]
target = ['yes','yes','yes','no']
final_hypothesis = finds(examples,target)
print("Final hypothesis : ",final_hypothesis)
```
