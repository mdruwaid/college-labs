```python
def candidate_eliminations(concepts, target):
    s = concepts[0].copy()
    G = [["?"] * len(s)]
    for i, h in enumerate(concepts):
        if target[i] == "yes":
            for x in range(len(s)):
                if h[x] != s[x]:
                    s[x] = "?"
            G = [
                g for g in G
                if all(g[x] == "?" or g[x] == h[x]
                       for x in range(len(s)))
            ]
        else:
            new_G = []
            for g in G:
                for x in range(len(s)):
                    if g[x] == "?":
                        if s[x] != "?" and s[x] != h[x]:
                            g_new = g.copy()
                            g_new[x] = s[x]
                            new_G.append(g_new)
            G = new_G
    return s, G

concepts = [
    ['sunny', 'warm', 'normal', 'strong', 'warn', 'same'],
    ['sunny', 'warm', 'high', 'strong', 'warn', 'same'],
    ['sunny', 'warm', 'high', 'strong', 'cool', 'change'],
    ['rainy', 'cold', 'high', 'strong', 'warn', 'change']
]

target = ['yes', 'yes', 'yes', 'no']
s_final, G_final = candidate_eliminations(concepts, target)
print("Final S:", s_final)
print("Final G:", G_final)
```
